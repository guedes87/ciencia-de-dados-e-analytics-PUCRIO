Pergunta 1: Qual a média de pontos para ser campeão?

```python
df = spark.sql(
    """
    SELECT ROUND(AVG(points)) AS media_pontos_campeao
    FROM brasileirao
    WHERE place = 1
    """
)
display(df)
```

Pergunta 2: Qual a média de pontos para ser rebaixado?

- Foi utilizado como base o primeiro time da zona de corte para o rebaixamento, o 17º da tabela.

```python
df = spark.sql(
    """
    SELECT ROUND(AVG(points)) AS media_pontos_rebaixado
    FROM brasileirao
    WHERE place = 17
    """
)
display(df)
```

Pergunta 3: Qual time com maior número de títulos?
- Apresenta a quantidade de títulos por time.

```python
df = spark.sql(
    """
    SELECT team, COUNT(*) as Qtd_titulos 
    FROM brasileirao
    WHERE place = 1
    GROUP BY team
    ORDER BY Qtd_titulos DESC
    """
)
display(df)
```

- Para complementar a análise, o código foi alterado para apresentar os anos em que o time foi campeão.

```python
df = spark.sql(
    """
    SELECT team, COUNT(*) as Qtd_titulos, COLLECT_LIST(season) as anos_titulos
    FROM brasileirao
    WHERE place = 1
    GROUP BY team
    ORDER BY Qtd_titulos DESC
    """
)
display(df)
```

Pergunta 4: Qual time com o maior número de gols marcados?

```python
df = spark.sql(
    """
    SELECT place, team, season, goals
    FROM brasileirao
    WHERE goals = (SELECT MAX(goals) FROM brasileirao)
    """)
display(df)
```

Pergunta 5:  Qual time com o menor número de gols marcados?

```python
df = spark.sql(
    """
    SELECT place, team, season, goals
    FROM brasileirao
    WHERE goals = (SELECT MIN(goals) FROM brasileirao)
    """)
display(df)
```
